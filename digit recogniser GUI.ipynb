```python
from tkinter import*
from tensorflow.keras.models import load_model
import numpy as np
from PIL import Image,ImageDraw
```

```python
model=load_model("digit_recogniser.h5")
```

```python


window=Tk()
window.title("Digit Recogniser")

def draw(event):
    x,y=event.x,event.y
    canvas.create_oval(x-3,y-3,x+3,y+3,fill="black",width=3)
    draw_object.ellipse([x-3,y-3,x+3,y+3],fill=0)

def clearing():
    global image,draw_object
    image=Image.new("L",(500,500),color="white")
    draw_object=ImageDraw.Draw(image)
    label.config(text="Draw a digit")
    canvas.delete("all")

def predict():
    resized=image.resize((28,28))
    resized=np.array(resized)/255
    img=np.expand_dims(resized,axis=0)
    prediction=model.predict(img)
    final=np.argmax(prediction)
    label.config(text=final)

#canvas
canvas=Canvas(window,width=500,height=500,bg="white")
canvas.grid(row=0,column=0,padx=3,pady=3)
canvas.bind("<B1-Motion>",draw)

label=Label(window,text="Draw a digit")
label.grid(row=0,column=1)

predict_button=Button(window,text="Predict",command=predict)
predict_button.grid(row=1,column=1)

clear_button=Button(window,text="clear",command=clearing)
clear_button.grid(row=1,column=0)
#creating image
image=Image.new("L",(500,500),color="white")
draw_object=ImageDraw.Draw(image)
window.mainloop()



```
