```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,Dense,SimpleRNN
```

```python
paragraph="Life is full of different experiences and opportunities. Every day, people wake up, get ready, and go about their daily activities. Some people enjoy reading books, while others prefer watching movies or listening to music. During the weekend, families often spend time together, visit new places, or go shopping. The weather can change throughout the day, bringing sunshine, rain, or cloudy skies. People learn new things, meet new friends, and develop different skills as they grow. Every experience teaches us something new and helps us understand the world around us."
words = sorted(set(paragraph.split()))
print(words)
```

```python
word_to_ind = {word: i for i, word in enumerate(words)}
print(word_to_ind)

ind_to_word = {i: word for i, word in enumerate(words)}
print(ind_to_word)

text_ind = [word_to_ind[word] for word in paragraph.split()]
print(text_ind)
```

```python
sequ_length = 4
x = []
y = []

for pos in range(0, len(text_ind) - sequ_length):
    x.append(text_ind[pos:pos+sequ_length])
    y.append(text_ind[pos+sequ_length])

x = np.array(x)
y = np.array(y)

print(x)
print(y)
```

```python
model = Sequential()

model.add(Embedding(input_dim=len(words),output_dim=50,input_length=sequ_length))

model.add(SimpleRNN(100))

model.add(Dense(len(words), activation='softmax'))

model.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(x, y, epochs=50, batch_size=20)
```

```python
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["Accuracy"])
```

```python
x_encoded=x
print(x_encoded)
```

```python
model.fit(x_encoded,y,epochs=50,batch_size=20)
```

```python
def predict(data):

    data = data.lower().split()

    if len(data) < sequ_length:
        data = [''] * (sequ_length - len(data)) + data

    elif len(data) > sequ_length:
        data = data[-sequ_length:]

    data_index = [word_to_ind[i] for i in data]

    encoded_data = np.array([data_index])

    prediction = model.predict(encoded_data)

    predicted_index = np.argmax(prediction, axis=-1)[0]

    final_prediction = ind_to_word[predicted_index]

    print(final_prediction)
```
