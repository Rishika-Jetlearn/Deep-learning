```python
!pip install tensorflow
import tensorflow as tf
from tensorflow.keras import layers,models
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
```

```python
(Xtr,ytr),(Xts,yts)=cifar10.load_data()
vehicle_classes=[0,1,8,9]
animal_classes=[2,3,4,5,6,7]
```

```python
print(ytr[:10])
```

```python
ytr=[(0 if i in vehicle_classes else 1) for i in ytr.flatten()]
```

```python
yts=[(0 if i in vehicle_classes else 1)for i in yts.flatten()]
```

```python
y_true=yts.copy()
print(ytr[:10])
```

```python
ytr=to_categorical(ytr,2)
yts=to_categorical(yts,2)
print(ytr.shape)
```

```python
Xtr=Xtr/255
Xts=Xts/255
print(Xtr.shape)
```

```python
cnn=models.Sequential([
    layers.Conv2D(32,(3,3),activation="relu",input_shape=(32,32,3)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64,(3,3),activation="relu"),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(128,(3,3),activation="relu"),
    layers.Flatten(),
    layers.Dense(64,activation="relu"),
    layers.Dense(2,activation="softmax")
])
cnn.compile(optimizer="Adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
history=cnn.fit(Xtr,ytr,epochs=10,validation_data=(Xts,yts))
```

```python
loss,accuracy=cnn.evaluate(Xts,yts)
print(loss)
print(accuracy)
```

```python
y_pred=cnn.predict(Xts)
print(y_pred)
```

```python
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import numpy as np

predicted_classes = np.argmax(y_pred,axis=1)
print(predicted_classes)
actual_classes=np.argmax(yts,axis=1)
print(actual_classes)

cm = confusion_matrix(actual_classes, predicted_classes)

ConfusionMatrixDisplay(cm,display_labels=["vehicle_classes","animal_classes"]).plot()

plt.show()
```

```python
def display_images(index,subplot_position):
  plt.subplot(2,3,subplot_position)
  plt.imshow(Xts[index])
  actual_label="vehicle" if actual_classes[index]==0 else "animal"
  predicted_label= "vehicle" if predicted_classes[index]==0 else "animal"
  plt.title(f"Actual:{actual_label},Predicted:{predicted_label}")
data=[18,234,5600,86,120,7]
plt.figure(figsize=(12,7))
for i,d in enumerate(data):
  display_images(d,i+1)
plt.show()
```
