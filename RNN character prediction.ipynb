```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN
```

```python
sentence="good morning"
characters=sorted(set(sentence))
print(characters)
```

```python
char_to_ind={ch:i for i,ch in enumerate(characters)}
print(char_to_ind)

ind_to_char={i:ch for i,ch in enumerate(characters)}
print(ind_to_char)

text_ind=[char_to_ind[ch] for ch in sentence]
print(text_ind)

```

```python
sequ_length=4
x=[]
y=[]

for pos in range(0,len(text_ind)-sequ_length):
    x.append(text_ind[pos:pos+sequ_length])
    y.append(text_ind[pos+sequ_length])

x=np.array(x)
y=np.array(y)

print(x)
print(y)
```

```python
model=Sequential()
model.add(SimpleRNN(128,input_shape=(sequ_length,len(characters)),activation="relu"))
model.add(Dense(len(characters),activation="softmax"))
```
