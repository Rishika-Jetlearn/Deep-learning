```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN
```

```python
sentence="good morning"
characters=sorted(set(sentence))
print(characters)
```

```python
char_to_ind={ch:i for i,ch in enumerate(characters)}
print(char_to_ind)

ind_to_char={i:ch for i,ch in enumerate(characters)}
print(ind_to_char)

text_ind=[char_to_ind[ch] for ch in sentence]
print(text_ind)

```

```python
sequ_length=4
x=[]
y=[]

for pos in range(0,len(text_ind)-sequ_length):
    x.append(text_ind[pos:pos+sequ_length])
    y.append(text_ind[pos+sequ_length])

x=np.array(x)
y=np.array(y)

print(x)
print(y)
```

```python
model=Sequential()
model.add(SimpleRNN(128,input_shape=(sequ_length,len(characters)),activation="relu"))
model.add(Dense(len(characters),activation="softmax"))
```

```python
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["Accuracy"])
```

```python
x_encoded=np.zeros((x.shape[0],x.shape[1],len(characters)))
#print(x_encoded)

for i,sequence in enumerate(x):
    for p,char in enumerate(sequence):
        x_encoded[i,p,char]=1

print(x_encoded)
```

```python
y=np.expand_dims(y,-1)
print(y)
```

```python
model.fit(x_encoded,y,epochs=50,batch_size=20)
```

```python
def predict(data):
    if len(data)<sequ_length:
        data=data.zfill(sequ_length)
    elif len(data)>sequ_length:
        data=data[-sequ_length:]

    data_index=[char_to_ind[i] for i in data]
    encoded_data=np.zeros((1,sequ_length,len(characters)))
    for i,p in enumerate(data_index):
        encoded_data[0,i,p]=1
    prediction=model.predict(encoded_data)

    predicted_index=np.argmax(prediction,-1)[0]
    final_prediction=ind_to_char[predicted_index]
    print(final_prediction)
```
