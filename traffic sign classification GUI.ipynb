```python
#GUI
from tkinter import*
from PIL import ImageTk, Image
from tkinter.filedialog import*
import numpy as np
```

```python
root=Tk()
root.geometry("500x500")
root.title("Traffic sign classifier")

traffic_class=Label(root)


def upload():
    traffic_class.pack_forget()
    try:
        file=askopenfilename()
        print(file)

        img=Image.open(file)

        img=img.resize((100,100))

        #display image:
        visImage = ImageTk.PhotoImage(img)

        image_label.config(image=visImage,text="")

        image_label.image=visImage
        classify(file)
    except:
        print("Image can't be imported")


def classify(path):
    img=Image.open(path)
    img=img.resize((30,30))
    img=np.array(img)
    img=img/255
    img=np.expand_dims(img,axis=0)
    prediction=model.predict(img)
    predicticed_class=np.argmax(prediction,axis=1)[0]+1
    print(predicticed_class)
    predicted_label=classes[predicticed_class]
    result.config(text=predicted_label)



title=Label(root, text="Traffic Light Detection Model",font=("Arial", 18, "bold"))
title.grid(row=0, column=0, columnspan=2, pady=20)

upload_label=Label(root, text="Uploaded Image",font=("Arial", 14))
upload_label.grid(row=1, column=0, padx=30)

dectection_label=Label(root, text="Detection Output",font=("Arial", 14))
dectection_label.grid(row=1, column=1, padx=30)

image_label=Label(root, text="[ Image Display ]", bg="lightgray")
image_label.grid(row=2, column=0, padx=20, pady=10)

result=Label(root, text="[ Results Display ]", bg="lightgray")
result.grid(row=2, column=1, padx=20, pady=10)

upload_button=Button(root, text="Upload Image",command=upload)
upload_button.grid(row=3, column=0, pady=15)

# detect_button=Button(root, text="Detect Traffic Light")
# detect_button.grid(row=3, column=1, pady=15)

mainloop()
```
